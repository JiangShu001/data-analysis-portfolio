```python
import pandas as pd

# 读取原始数据
df = pd.read_csv('../data/raw/UserBehavior.csv', header=None,
                 names=['用户ID', '商品ID', '类目ID', '行为类型', '时间戳'])

print("清洗前：")
print(f"行数：{len(df)}")

# 1. 删除缺失值
df = df.dropna()

# 2. 删除重复行
df = df.drop_duplicates()

# 3. 时间戳转日期（用pandas，不用time）
df['日期'] = pd.to_datetime(df['时间戳'], unit='s').dt.strftime('%Y-%m-%d')

# 4. 过滤日期范围
df = df[(df['日期'] >= '2017-11-25') & (df['日期'] <= '2017-12-03')]

# 5. 过滤异常行为类型
df = df[df['行为类型'].isin(['pv', 'cart', 'fav', 'buy'])]

print("\n清洗后：")
print(f"行数：{len(df)}")
print(f"日期范围：{df['日期'].min()} 至 {df['日期'].max()}")
print(f"行为类型分布：\n{df['行为类型'].value_counts()}")

# 保存清洗后数据
df.to_csv('../data/cleaned/user_behavior_cleaned.csv', index=False)
print("\n已保存到 ../data/cleaned/user_behavior_cleaned.csv")
```
