```python
import pandas as pd
import numpy as np
import os

# 固定随机种子，保证结果可复现
np.random.seed(42)

# 生成日期范围：2024年1月1日到3月31日，共91天
dates = pd.date_range('2024-01-01', '2024-03-31', freq='D')

# 定义5个渠道，每个渠道有不同的"性格"
channels = {
    '小红书': {'曝光基数': 50000, 'ctr': 0.045, 'cvr': 0.030, 'cpc': 1.2},
    '抖音':   {'曝光基数': 80000, 'ctr': 0.035, 'cvr': 0.020, 'cpc': 0.9},
    '百度':   {'曝光基数': 60000, 'ctr': 0.025, 'cvr': 0.015, 'cpc': 1.5},
    '微信':   {'曝光基数': 40000, 'ctr': 0.050, 'cvr': 0.035, 'cpc': 1.8},
    'B站':    {'曝光基数': 30000, 'ctr': 0.030, 'cvr': 0.010, 'cpc': 0.7},
}

rows = []
for date in dates:
    for ch, p in channels.items():
        exposure = int(p['曝光基数'] * np.random.uniform(0.8, 1.2))
        clicks = int(exposure * p['ctr'] * np.random.uniform(0.85, 1.15))
        conversions = int(clicks * p['cvr'] * np.random.uniform(0.8, 1.2))
        cost = round(clicks * p['cpc'] * np.random.uniform(0.9, 1.1), 2)
        revenue = round(conversions * 200 * np.random.uniform(0.9, 1.1), 2)

        rows.append({
            '日期': date,
            '渠道': ch,
            '曝光量': exposure,
            '点击量': clicks,
            '转化量': conversions,
            '花费': cost,
            '收入': revenue
        })

df = pd.DataFrame(rows)

# 保存到 data/raw/，用相对路径
df.to_csv('../data/raw/channel_data.csv', index=False, encoding='utf-8-sig')

print("数据形状：", df.shape)
print("保存路径：", os.path.abspath('../data/raw/channel_data.csv'))
df.head(10)
```
