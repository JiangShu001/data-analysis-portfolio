```python
import pandas as pd
import numpy as np

# 加载数据
df = pd.read_csv('../data/raw/channel_data.csv')

# 看基本信息
print("形状：", df.shape)
print("\n列名：", df.columns.tolist())
print("\n数据类型：")
print(df.dtypes)
print("\n前5行：")
df.head()
```

```python
# 把日期列转成datetime类型
df['日期'] = pd.to_datetime(df['日期'])

# 再检查一次类型
print(df.dtypes)
```

```python
print("缺失值：")
print(df.isnull().sum())

print("\n重复行数：", df.duplicated().sum())

print("\n数值列描述统计：")
print(df[['曝光量', '点击量', '转化量', '花费', '收入']].describe())
```

```python
# 保存清洗后的数据
df.to_csv('../data/cleaned/channel_clean.csv', index=False, encoding='utf-8-sig')

print("已保存到 data/cleaned/channel_clean.csv")
print("形状：", df.shape)
```
