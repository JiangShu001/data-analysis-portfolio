```python
import pandas as pd

# 加载清洗后的数据
df = pd.read_csv('../data/cleaned/channel_clean.csv')

# 按渠道汇总
channel_summary = df.groupby('渠道').agg({
    '曝光量': 'sum',
    '点击量': 'sum',
    '转化量': 'sum',
    '花费': 'sum',
    '收入': 'sum'
}).reset_index()

# 算比率指标
channel_summary['CTR'] = channel_summary['点击量'] / channel_summary['曝光量']
channel_summary['CVR'] = channel_summary['转化量'] / channel_summary['点击量']
channel_summary['ROI'] = channel_summary['收入'] / channel_summary['花费']
channel_summary['客单价'] = channel_summary['收入'] / channel_summary['转化量']

# 按ROI降序排列
channel_summary = channel_summary.sort_values('ROI', ascending=False)

# 显示
print(channel_summary.to_string(index=False))
```

```python
channel_summary.to_csv('../data/cleaned/channel_summary.csv', index=False, encoding='utf-8-sig')
print("已保存")
```
